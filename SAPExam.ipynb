```python
"""
SAP Mock Exam - Chapters 3, 4, and 5
Total Questions: 60
Run this file in VS Code or any Python environment
"""

import random
import json
from datetime import datetime

class MockExam:
    def __init__(self):
        self.questions = [
            # CHAPTER 3 - Financial Accounting (20 questions)
            {
                "id": 1,
                "chapter": "Chapter 3",
                "question": "Which of the following is a reconciliation account in the general ledger?",
                "options": [
                    "Customer account",
                    "Supplies expense account",
                    "Accounts receivable account",
                    "Common stock account"
                ],
                "correct": [2],
                "multiple": False
            },
            {
                "id": 2,
                "chapter": "Chapter 3",
                "question": "Which of the following statements concerning reconciliation accounts are true? (2 answers)",
                "options": [
                    "The same reconciliation account can be used for both customers and vendors",
                    "They are specified in the definition of the associated sub-ledger account",
                    "The reconciliation account for customers is the accounts receivable account",
                    "It is possible to make direct postings to a reconciliation account"
                ],
                "correct": [1, 2],
                "multiple": True
            },
            {
                "id": 3,
                "chapter": "Chapter 3",
                "question": "Which of the following statements about reconciliation accounts is NOT true?",
                "options": [
                    "It is possible to make direct postings to a reconciliation account",
                    "Reconciliation accounts include totals of the associated sub ledger",
                    "Reconciliation accounts are part of the general ledger",
                    "Reconciliation accounts are specified in the definition of the associated sub ledger account"
                ],
                "correct": [0],
                "multiple": False
            },
            {
                "id": 4,
                "chapter": "Chapter 3",
                "question": "Which of the following are correct FI postings to record the sale of bicycles to a customer on credit? (3 answers)",
                "options": [
                    "Debit accounts receivable reconciliation account",
                    "Credit customer subledger account",
                    "Credit revenue account",
                    "Debit revenue account",
                    "Credit accounts receivable reconciliation account",
                    "Debit customer subledger account"
                ],
                "correct": [0, 2, 5],
                "multiple": True
            },
            {
                "id": 5,
                "chapter": "Chapter 3",
                "question": "Which of the following types of charts of accounts (CoA) is the primary chart of account?",
                "options": [
                    "Country-specific CoA",
                    "Group CoA",
                    "Operative CoA",
                    "Company code CoA"
                ],
                "correct": [2],
                "multiple": False
            },
            {
                "id": 6,
                "chapter": "Chapter 3",
                "question": "Which of the following is not a subledger account?",
                "options": [
                    "Vendor account",
                    "Specific asset account",
                    "Accounts receivable account",
                    "Customer account"
                ],
                "correct": [2],
                "multiple": False
            },
            {
                "id": 7,
                "chapter": "Chapter 3",
                "question": "Which of the following statements about a company code are true? (2 answers)",
                "options": [
                    "A client may only have one company code",
                    "All data needed to prepare reports for internal use (to manage the company) are maintained at the company code level",
                    "All data needed to prepare reports for external reporting are maintained at the company code level",
                    "Each company code represents a separate legal entity within an enterprise"
                ],
                "correct": [2, 3],
                "multiple": True
            },
            {
                "id": 8,
                "chapter": "Chapter 3",
                "question": "Which of the following statements about FI and CO is false?",
                "options": [
                    "Management accounting focuses on managing processes that help achieve basic business objectives related to costs and revenues",
                    "FI is Externally focused while CO is internally focused",
                    "FI reports are based on management's needs",
                    "Accounting processes are broadly divided into FI and CO"
                ],
                "correct": [2],
                "multiple": False
            },
            {
                "id": 9,
                "chapter": "Chapter 3",
                "question": "Which of the following statements about FI and CO is true?",
                "options": [
                    "FI reports are primarily intended to meet legal or regulatory requirements",
                    "FI reports are not relevant for internal management purposes",
                    "CO reports are primarily intended to meet legal or regulatory requirements",
                    "FI reports focus on costs and revenues"
                ],
                "correct": [0],
                "multiple": False
            },
            {
                "id": 10,
                "chapter": "Chapter 3",
                "question": "Which of the following is a balance sheet account?",
                "options": [
                    "Rent expense account",
                    "Supplies expense account",
                    "Accounts payable account",
                    "Cost of goods sold account"
                ],
                "correct": [2],
                "multiple": False
            },
            {
                "id": 11,
                "chapter": "Chapter 3",
                "question": "General ledger accounts include which of the following segments of data? (2 answers)",
                "options": [
                    "Company code segment",
                    "Chart of accounts segment",
                    "Vendor segment",
                    "General ledger segment"
                ],
                "correct": [0, 1],
                "multiple": True
            },
            {
                "id": 12,
                "chapter": "Chapter 3",
                "question": "Which of the following are correct FI postings to record payment to a vendor for supplies previously purchased on credit? (3 answers)",
                "options": [
                    "Debit vendor subledger account",
                    "Credit bank account",
                    "Debit accounts payable reconciliation account",
                    "Credit accounts payable reconciliation account",
                    "Debit bank account",
                    "Credit vendor subledger account"
                ],
                "correct": [0, 1, 2],
                "multiple": True
            },
            {
                "id": 13,
                "chapter": "Chapter 3",
                "question": "General ledger postings include a debit entry and a credit entry. Which of the following will result in a debit entry? (2 answers)",
                "options": [
                    "When there is an increase in assets",
                    "When there is a decrease in assets",
                    "When there is a decrease in expenses",
                    "When there is an increase in expenses"
                ],
                "correct": [0, 3],
                "multiple": True
            },
            {
                "id": 14,
                "chapter": "Chapter 3",
                "question": "Which of the following is master data in financial accounting?",
                "options": [
                    "Vendor Subledger",
                    "Company code",
                    "Income statement",
                    "Balance Sheet"
                ],
                "correct": [0],
                "multiple": False
            },
            {
                "id": 15,
                "chapter": "Chapter 3",
                "question": "Which of the following statements about balance sheets are true? (2 answers)",
                "options": [
                    "A balance sheet indicates changes in a company's financial position over a period of time",
                    "A balance sheet include revenue and expenses",
                    "A balance sheet include assets, liabilities and equity",
                    "A balance sheet is a snapshot of an organization at a point in time"
                ],
                "correct": [2, 3],
                "multiple": True
            },
            {
                "id": 16,
                "chapter": "Chapter 3",
                "question": "Which of the following is a profit and loss account?",
                "options": [
                    "Rent expense account",
                    "Prepaid rent account",
                    "Fixed asset account",
                    "Accounts receivable account"
                ],
                "correct": [0],
                "multiple": False
            },
            {
                "id": 17,
                "chapter": "Chapter 3",
                "question": "Which of the following is an equity account in the general ledger?",
                "options": [
                    "Inventory account",
                    "Retained earnings",
                    "Accounts payable account",
                    "Accounts receivable account"
                ],
                "correct": [1],
                "multiple": False
            },
            {
                "id": 18,
                "chapter": "Chapter 3",
                "question": "Which of the following is the correct FI posting for the following scenario: GBI received $10,000 from an investor in exchange for common stock in the company?",
                "options": [
                    "Debit bank account, credit common stock account",
                    "Debit bank account, credit revenue account",
                    "Debit bank account, credit accounts payable",
                    "Credit bank account, debit common stock account"
                ],
                "correct": [0],
                "multiple": False
            },
            {
                "id": 19,
                "chapter": "Chapter 3",
                "question": "GBI purchases $2000 of office supplies on credit with payment to be made at a later date. Which of the following is the correct posting at the time of purchase? (2 answers)",
                "options": [
                    "Credit supplies expenses",
                    "Credit bank",
                    "Debit supplies expenses",
                    "Credit payable"
                ],
                "correct": [2, 3],
                "multiple": True
            },
            {
                "id": 20,
                "chapter": "Chapter 3",
                "question": "Which of the following financial accounting processes is used to record data related to purchase, use and disposal of machinery, buildings etc?",
                "options": [
                    "Management accounting",
                    "Accounts payable accounting",
                    "Asset accounting",
                    "General ledger accounting"
                ],
                "correct": [2],
                "multiple": False
            },

            # CHAPTER 4 - Procurement Process (20 questions)
            {
                "id": 21,
                "chapter": "Chapter 4",
                "question": "Purchasing data in a vendor master is defined for which organizational level?",
                "options": [
                    "Company code",
                    "Purchasing organization",
                    "Client",
                    "Plant"
                ],
                "correct": [1],
                "multiple": False
            },
            {
                "id": 22,
                "chapter": "Chapter 4",
                "question": "Which of the following are master data relevant to the Procurement process? (2 answers)",
                "options": [
                    "Vendor",
                    "Storage location",
                    "Plant",
                    "Purchasing info Records/Conditions"
                ],
                "correct": [0, 3],
                "multiple": True
            },
            {
                "id": 23,
                "chapter": "Chapter 4",
                "question": "Which of the following general ledger accounts are relevant to the procurement process? (2 answers)",
                "options": [
                    "Vendor account",
                    "Goods receipt / Invoice receipt (GR/IR) account",
                    "Accounts payable"
                ],
                "correct": [1, 2],
                "multiple": True
            },
            {
                "id": 24,
                "chapter": "Chapter 4",
                "question": "Which of the following are segments of data in a vendor master? (2 answers)",
                "options": [
                    "General Data",
                    "Accounting Data",
                    "Purchasing Data",
                    "Sales Data"
                ],
                "correct": [0, 2],
                "multiple": True
            },
            {
                "id": 25,
                "chapter": "Chapter 4",
                "question": "Which of the following are organizational data relevant to the procurement process? (2 answers)",
                "options": [
                    "Purchasing organization",
                    "Storage location",
                    "Purchase requisition",
                    "Vendor"
                ],
                "correct": [0, 1],
                "multiple": True
            },
            {
                "id": 26,
                "chapter": "Chapter 4",
                "question": "Basic data in a vendor master is defined for which organizational level?",
                "options": [
                    "Company code",
                    "Client",
                    "Purchasing organization",
                    "Plant"
                ],
                "correct": [1],
                "multiple": False
            },
            {
                "id": 27,
                "chapter": "Chapter 4",
                "question": "Which of the following data in the material master are updated during the procurement process?",
                "options": [
                    "Currency",
                    "Purchasing group",
                    "Inventory value/Quantity in Stock",
                    "Purchasing organization"
                ],
                "correct": [2],
                "multiple": False
            },
            {
                "id": 28,
                "chapter": "Chapter 4",
                "question": "In which of the following steps in the procurement process can inventory value be updated? (2 answers)",
                "options": [
                    "Receive invoice",
                    "Create purchase requisition",
                    "Send payment",
                    "Receive Materials"
                ],
                "correct": [2, 3],
                "multiple": True
            },
            {
                "id": 29,
                "chapter": "Chapter 4",
                "question": "Which of the following subledger accounts are relevant to the procurement process?",
                "options": [
                    "Customer account",
                    "Vendor account",
                    "GR/IR account",
                    "Inventory account"
                ],
                "correct": [1],
                "multiple": False
            },
            {
                "id": 30,
                "chapter": "Chapter 4",
                "question": "In which of the following steps of the procurement process is the GR/IR account updated? (2 answers)",
                "options": [
                    "Receive materials (goods receipt)",
                    "Receive invoice",
                    "Create purchase requisition",
                    "Create purchase order",
                    "Send payment"
                ],
                "correct": [0, 1],
                "multiple": True
            },
            {
                "id": 31,
                "chapter": "Chapter 4",
                "question": "Which of the following data segments (views) of the material master are relevant to the purchasing process? (2 answers)",
                "options": [
                    "Vendor data",
                    "Customer data",
                    "Plant data",
                    "Purchasing data",
                    "Accounting data"
                ],
                "correct": [3, 4],
                "multiple": True
            },
            {
                "id": 32,
                "chapter": "Chapter 4",
                "question": "Which is the first step in the procurement process that has a financial accounting impact?",
                "options": [
                    "Send payment",
                    "Receive invoice",
                    "Receive materials (goods receipt)",
                    "Create purchase requisition",
                    "Create purchase order"
                ],
                "correct": [2],
                "multiple": False
            },
            {
                "id": 33,
                "chapter": "Chapter 4",
                "question": "Which of the following are the consequences (outcomes) of completing a goods receipt step? (2 answers)",
                "options": [
                    "A material document is created",
                    "The purchasing requisition is updated",
                    "The purchase price is updated",
                    "An FI document is created"
                ],
                "correct": [0, 3],
                "multiple": True
            },
            {
                "id": 34,
                "chapter": "Chapter 4",
                "question": "From which of the following stock types can inventory be withdrawn for consumption?",
                "options": [
                    "Blocked stock",
                    "Unrestricted use",
                    "Stock in transit"
                ],
                "correct": [1],
                "multiple": False
            },
            {
                "id": 35,
                "chapter": "Chapter 4",
                "question": "Purchasing info records include data from which of the following master data? (2 answers)",
                "options": [
                    "Storage location",
                    "Company code",
                    "Vendor",
                    "Material"
                ],
                "correct": [2, 3],
                "multiple": True
            },
            {
                "id": 36,
                "chapter": "Chapter 4",
                "question": "Accounting data in a vendor master is defined for which organizational level?",
                "options": [
                    "Purchasing group",
                    "Client",
                    "Storage location",
                    "Company code"
                ],
                "correct": [3],
                "multiple": False
            },
            {
                "id": 37,
                "chapter": "Chapter 4",
                "question": "Which step immediately follows the 'create purchase order' step in a standard procurement process?",
                "options": [
                    "Send payment",
                    "Receive invoice",
                    "Receive materials (goods receipt)",
                    "Create purchase requisition"
                ],
                "correct": [2],
                "multiple": False
            },
            {
                "id": 38,
                "chapter": "Chapter 4",
                "question": "Which of the following are item categories in a purchase order? (2 answers)",
                "options": [
                    "Consignment",
                    "Quality inspection",
                    "Subcontracting",
                    "Unrestricted use"
                ],
                "correct": [0, 2],
                "multiple": True
            },
            {
                "id": 39,
                "chapter": "Chapter 4",
                "question": "Which of the following steps in the procurement process does not have a financial accounting impact?",
                "options": [
                    "Receive materials (goods receipt)",
                    "Create purchase requisition",
                    "Receive invoice",
                    "Send payment"
                ],
                "correct": [1],
                "multiple": False
            },
            {
                "id": 40,
                "chapter": "Chapter 4",
                "question": "In which of the following steps of the procurement process is the bank account updated?",
                "options": [
                    "Receive materials (goods receipt)",
                    "Create purchase requisition",
                    "Receive invoice",
                    "Create purchase order",
                    "Send payment"
                ],
                "correct": [4],
                "multiple": False
            },

            # CHAPTER 5 - Fulfillment Process (20 questions)
            {
                "id": 41,
                "chapter": "Chapter 5",
                "question": "Customer master data is defined for which of the following organization levels? (2 answers)",
                "options": [
                    "Sales area",
                    "Distribution chain",
                    "Company code",
                    "Division"
                ],
                "correct": [0, 2],
                "multiple": True
            },
            {
                "id": 42,
                "chapter": "Chapter 5",
                "question": "In which of the following steps of the fulfillment process is a material document created?",
                "options": [
                    "Send shipment (goods issue)",
                    "Receive materials (goods receipt)",
                    "Create purchase order",
                    "Receive customer payment"
                ],
                "correct": [0],
                "multiple": False
            },
            {
                "id": 43,
                "chapter": "Chapter 5",
                "question": "A customer-material information record includes data specific to which of the following?",
                "options": [
                    "One customer and one plant",
                    "One customer and one material",
                    "All materials in one plant",
                    "One material and one plant"
                ],
                "correct": [1],
                "multiple": False
            },
            {
                "id": 44,
                "chapter": "Chapter 5",
                "question": "Which of the following steps in the fulfillment process have a financial accounting impact? (2 answers)",
                "options": [
                    "Send customer invoice",
                    "Receive customer payment",
                    "Create quote",
                    "Create sales order",
                    "Ship materials"
                ],
                "correct": [0, 1],
                "multiple": True
            },
            {
                "id": 45,
                "chapter": "Chapter 5",
                "question": "Basic data in the material master is defined for which of the following organizational level?",
                "options": [
                    "Division",
                    "Sales organization",
                    "Plant",
                    "Client"
                ],
                "correct": [3],
                "multiple": False
            },
            {
                "id": 46,
                "chapter": "Chapter 5",
                "question": "Which of the following are typical distribution channels? (2 Answers)",
                "options": [
                    "Wholesale",
                    "Online",
                    "Warehouse",
                    "Distribution chain"
                ],
                "correct": [0, 1],
                "multiple": True
            },
            {
                "id": 47,
                "chapter": "Chapter 5",
                "question": "Which of the following is an outcome of completing the billing step of the fulfillment process?",
                "options": [
                    "Inventory account is debited",
                    "Accounts receivable reconciliation account is credited",
                    "Customer sub ledger account is debited",
                    "Sales revenue account is debited"
                ],
                "correct": [2],
                "multiple": False
            },
            {
                "id": 48,
                "chapter": "Chapter 5",
                "question": "Which of the following is not a valid reference document that can be used to create a quotation?",
                "options": [
                    "Quotation",
                    "Sales order",
                    "Agreements",
                    "FI document"
                ],
                "correct": [3],
                "multiple": False
            },
            {
                "id": 49,
                "chapter": "Chapter 5",
                "question": "Which of the following reference documents can be used to create a quotation? (2 Answers)",
                "options": [
                    "Customer inquiry",
                    "Material document",
                    "Sales order",
                    "Customer Invoice"
                ],
                "correct": [0, 2],
                "multiple": True
            },
            {
                "id": 50,
                "chapter": "Chapter 5",
                "question": "Which of the following master data is necessary to create a quote in the fulfillment process?",
                "options": [
                    "Plant",
                    "Client",
                    "Material master",
                    "Sales area"
                ],
                "correct": [2],
                "multiple": False
            },
            {
                "id": 51,
                "chapter": "Chapter 5",
                "question": "Which of the following is NOT a typical distribution channel?",
                "options": [
                    "Retail",
                    "Online",
                    "Wholesale",
                    "Distribution chain"
                ],
                "correct": [3],
                "multiple": False
            },
            {
                "id": 52,
                "chapter": "Chapter 5",
                "question": "Which of the following sub-ledger accounts is credited when a customer payment is recorded?",
                "options": [
                    "Inventory account",
                    "Vendor account",
                    "Customer account",
                    "Revenue account"
                ],
                "correct": [2],
                "multiple": False
            },
            {
                "id": 53,
                "chapter": "Chapter 5",
                "question": "Which of the following statements about the relationship between divisions and sales organizations is true? (2 Answers)",
                "options": [
                    "A sales organization can have a maximum of two divisions",
                    "A sales organization need not have a division assigned to it",
                    "A division can be assigned to multiple sales organizations",
                    "Many divisions can be assigned to a sales organization"
                ],
                "correct": [2, 3],
                "multiple": True
            },
            {
                "id": 54,
                "chapter": "Chapter 5",
                "question": "Which of the following organizational data is necessary to create a quote in the fulfillment process?",
                "options": [
                    "Company code",
                    "Shipping point",
                    "Plant",
                    "Customer master"
                ],
                "correct": [3],
                "multiple": False
            },
            {
                "id": 55,
                "chapter": "Chapter 5",
                "question": "Which of the following organizational data are needed to create an outbound delivery in shipping? (2 answers)",
                "options": [
                    "Delivering plant",
                    "Storage location",
                    "Sales area",
                    "Picking document"
                ],
                "correct": [0, 1],
                "multiple": True
            },
            {
                "id": 56,
                "chapter": "Chapter 5",
                "question": "Which of the following master data do companies use to determine the selling price of their products?",
                "options": [
                    "Material master",
                    "Vendor master",
                    "Output conditions",
                    "Pricing conditions"
                ],
                "correct": [3],
                "multiple": False
            },
            {
                "id": 57,
                "chapter": "Chapter 5",
                "question": "Which of the following is an outcome of creating a sales order?",
                "options": [
                    "An CO document is created",
                    "A transaction document is created",
                    "A material document is created",
                    "An FI document is created"
                ],
                "correct": [1],
                "multiple": False
            },
            {
                "id": 58,
                "chapter": "Chapter 5",
                "question": "Which of the following are examples of shipping points? (2 answers)",
                "options": [
                    "Plant",
                    "A nearby post office",
                    "Rail depot",
                    "Designated employees"
                ],
                "correct": [0, 2],
                "multiple": True
            },
            {
                "id": 59,
                "chapter": "Chapter 5",
                "question": "Which of the following master data are relevant to the fulfillment process? (2 answers)",
                "options": [
                    "Pricing conditions",
                    "Vendor master",
                    "Customer-material info record",
                    "Purchasing info record"
                ],
                "correct": [0, 2],
                "multiple": True
            },
            {
                "id": 60,
                "chapter": "Chapter 5",
                "question": "Which of the following is an outcome of recording a goods issue in the fulfillment process?",
                "options": [
                    "Bank account is credited",
                    "Inventory account is credited and cost of goods sold account is debited",
                    "Revenue account is debited",
                    "Inventory account is debited and cost of goods sold account is credited"
                ],
                "correct": [1],
                "multiple": False
            }
        ]

        self.user_answers = {}
        self.score = 0

    def display_question(self, q_num):
        """Display a single question"""
        q = self.questions[q_num]
        print(f"\n{'='*80}")
        print(f"Question {q['id']} ({q['chapter']})")
        print(f"{'='*80}")
        print(f"\n{q['question']}\n")

        for i, option in enumerate(q['options']):
            print(f"{chr(97+i)}) {option}")

        if q['multiple']:
            print("\n(Multiple answers - enter letters separated by commas, e.g., a,c,d)")

        return q

    def get_user_answer(self, question):
        """Get and validate user input"""
        while True:
            try:
                answer = input("\nYour answer: ").strip().lower()

                if not answer:
                    print("Please enter an answer.")
                    continue

                if question['multiple']:
                    # Handle multiple answers
                    selected = [ord(a.strip()) - 97 for a in answer.split(',')]
                    if all(0 <= s < len(question['options']) for s in selected):
                        return sorted(selected)
                    else:
                        print("Invalid selection. Please try again.")
                else:
                    # Handle single answer
                    selected = ord(answer[0]) - 97
                    if 0 <= selected < len(question['options']):
                        return [selected]
                    else:
                        print("Invalid selection. Please try again.")
            except:
                print("Invalid input. Please try again.")

    def check_answer(self, question, user_answer):
        """Check if answer is correct"""
        return sorted(user_answer) == sorted(question['correct'])

    def show_correct_answer(self, question):
        """Display the correct answer"""
        correct_letters = [chr(97 + i) for i in question['correct']]
        correct_options = [question['options'][i] for i in question['correct']]

        print("\n" + "="*80)
        print("CORRECT ANSWER:")
        print("="*80)
        for letter, option in zip(correct_letters, correct_options):
            print(f"{letter}) {option}")

    def run_exam(self, show_answers=False, randomize=False):
        """Run the complete exam"""
        print("\n" + "="*80)
        print("SAP MOCK EXAM - CHAPTERS 3, 4, AND 5")
        print("="*80)
        print(f"Total Questions: {len(self.questions)}")
        print("="*80)

        input("\nPress Enter to start the exam...")

        questions_order = list(range(len(self.questions)))
        if randomize:
            random.shuffle(questions_order)

        for idx, q_num in enumerate(questions_order):
            question = self.display_question(q_num)
            user_answer = self.get_user_answer(question)
            self.user_answers[question['id']] = user_answer

            is_correct = self.check_answer(question, user_answer)

            if is_correct:
                self.score += 1
                print("\n✓ Correct!")
            else:
                print("\n✗ Incorrect!")
                if show_answers:
                    self.show_correct_answer(question)

            # Show progress
            print(f"\nProgress: {idx + 1}/{len(self.questions)} | Current Score: {self.score}/{idx + 1}")

            if idx < len(self.questions) - 1:
                input("\nPress Enter for next question...")

        self.show_results()

    def show_results(self):
        """Display final results"""
        percentage = (self.score / len(self.questions)) * 100

        print("\n" + "="*80)
        print("EXAM RESULTS")
        print("="*80)
        print(f"\nTotal Questions: {len(self.questions)}")
        print(f"Correct Answers: {self.score}")
        print(f"Incorrect Answers: {len(self.questions) - self.score}")
        print(f"Score: {percentage:.2f}%")

        # Grade
        if percentage >= 90:
            grade = "A (Excellent!)"
        elif percentage >= 80:
            grade = "B (Very Good)"
        elif percentage >= 70:
            grade = "C (Good)"
        elif percentage >= 60:
            grade = "D (Pass)"
        else:
            grade = "F (Fail)"

        print(f"Grade: {grade}")
        print("="*80)

        # Save results
        self.save_results(percentage)

    def save_results(self, percentage):
        """Save exam results to a file"""
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        result_data = {
            "timestamp": timestamp,
            "total_questions": len(self.questions),
            "correct_answers": self.score,
            "percentage": percentage,
            "answers": self.user_answers
        }

        filename = f"exam_results_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"

        try:
            with open(filename, 'w') as f:
                json.dump(result_data, f, indent=4)
            print(f"\nResults saved to: {filename}")
        except Exception as e:
            print(f"\nCould not save results: {e}")

    def review_answers(self):
        """Review all questions with correct answers"""
        print("\n" + "="*80)
        print("ANSWER KEY - ALL QUESTIONS")
        print("="*80)

        for i, question in enumerate(self.questions):
            print(f"\n{'='*80}")
            print(f"Question {question['id']} ({question['chapter']})")
            print(f"{'='*80}")
            print(f"\n{question['question']}\n")

            for j, option in enumerate(question['options']):
                marker = "✓" if j in question['correct'] else " "
                print(f"[{marker}] {chr(97+j)}) {option}")

            # Show user's answer if available
            if question['id'] in self.user_answers:
                user_ans = self.user_answers[question['id']]
                user_letters = [chr(97 + i) for i in user_ans]
                correct = sorted(user_ans) == sorted(question['correct'])
                status = "✓ CORRECT" if correct else "✗ INCORRECT"
                print(f"\nYour answer: {', '.join(user_letters)} - {status}")

            if i < len(self.questions) - 1:
                input("\nPress Enter for next question...")


def main():
    """Main function to run the exam"""
    print("\n" + "="*80)
    print("SAP MOCK EXAM - CHAPTERS 3, 4, AND 5")
    print("="*80)
    print("\nWelcome to the SAP Mock Exam!")
    print("\nThis exam covers:")
    print("  - Chapter 3: Financial Accounting (20 questions)")
    print("  - Chapter 4: Procurement Process (20 questions)")
    print("  - Chapter 5: Fulfillment Process (20 questions)")
    print("\nTotal: 60 Questions")
    print("="*80)

    exam = MockExam()

    while True:
        print("\nChoose an option:")
        print("1. Take the exam (Show correct answers after each question)")
        print("2. Take the exam (See results only at the end)")
        print("3. Take the exam with randomized questions")
        print("4. View answer key (All questions with answers)")
        print("5. Exit")

        choice = input("\nEnter your choice (1-5): ").strip()

        if choice == '1':
            exam = MockExam()
            exam.run_exam(show_answers=True, randomize=False)

            review = input("\nWould you like to review all answers? (y/n): ").strip().lower()
            if review == 'y':
                exam.review_answers()

        elif choice == '2':
            exam = MockExam()
            exam.run_exam(show_answers=False, randomize=False)

            review = input("\nWould you like to review all answers? (y/n): ").strip().lower()
            if review == 'y':
                exam.review_answers()

        elif choice == '3':
            exam = MockExam()
            exam.run_exam(show_answers=True, randomize=True)

            review = input("\nWould you like to review all answers? (y/n): ").strip().lower()
            if review == 'y':
                exam.review_answers()

        elif choice == '4':
            exam.review_answers()

        elif choice == '5':
            print("\nThank you for using the SAP Mock Exam!")
            print("Good luck with your studies!")
            break

        else:
            print("\nInvalid choice. Please try again.")


if __name__ == "__main__":
    main()
```
